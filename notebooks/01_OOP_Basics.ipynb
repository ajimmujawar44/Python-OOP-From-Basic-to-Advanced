## 1. What is OOP?

**OOP = Object-Oriented Programming**

OOP is a programming style where we organize a program around **objects**.

An object contains:

- **Data** → attributes/properties
- **Behavior** → methods/actions

### Real-world example

Think about a `Car`.

**Data:**
- brand
- model
- color
- speed

**Behavior:**
- start()
- stop()
- accelerate()
- brake()

In OOP:

```text
Class  → Blueprint
Object → Real instance created from the blueprint
```

For example:

```text
Class: Car

Objects:
    car1 → BMW
    car2 → Toyota
    car3 → Audi
```

## 2. Why Do We Need OOP?

Without OOP, a large program can become difficult to organize because data and functions may be scattered.

For example:

```python
student_name = "Ajim"
student_age = 30
student_marks = 85

def display_student():
    ...
```

With OOP, related data and behavior can be grouped together:

```text
Student
 ├── name
 ├── age
 ├── marks
 ├── display()
 └── result()
```

### Main benefits

1. **Organization** — related data and behavior stay together.
2. **Reusability** — one class can create many objects.
3. **Maintainability** — large programs become easier to manage.
4. **Extensibility** — inheritance and polymorphism make systems easier to extend.
5. **Real-world modeling** — objects can represent things such as users, products, accounts, and ML models.

## 3. Class

A **class** is a blueprint or template for creating objects.

### Syntax

```python

    # attributes
    # methods
    pass
```

Let's create a simple `Student` class.

````python
class Student :
    pass
````

The class above does not contain any data or behavior yet.

pass means :

But we can already create objects from this class.

## 4. Object

An **objects** is an instance of a  class.

### Syntax 

'''python

'''


````python
class Student:
    pass

student1= Student()
print(type(student1))
````

### What happened?

```python

```

creates an object from the `Student` class.

Conceptually:

```text
Student (Class)
      |

```

`type(student1)` tells us that `student1` is a `Student` object.

## 6. Class vs Object

| Class | Object |
|---|---|
| Blueprint/template | Actual instance |
| Defines structure and behavior | Uses that structure and behavior |
| Does not represent one specific instance | Represents a specific instance |
| Example: `Student` | Example: `student1` |
| Can create many objects | Belongs to a class |

### Easy analogy

```text
Class  → House blueprint
Object → Actual house
```

## 7. Adding Attributes

Attributes are data/properties associated with an object.

We can add attributes directly to an object.

````python
class Student:
    pass
student1 = Student()

student1.name = "Azeem"
student1.age = 30
student1.course = "Data Science"

print(student1.name)
print(student1.age)
print(student1.course)
````

However, assigning attributes manually like this is not the best way to design a class. Later we will use `__init__()` to initialize object data cleanly.

## 8. Attributes Belong to Individual Objects

Different objects can have different values.

````python
class Student:
    pass

student1 = Student()

student2 = Student()

student1.name = "Azeem"
student1.age = 28

student2.name = "Aryan"
student2.age = 25

print(student1.name, student1.age)
print(student2.name, student2.age)
````

Notice:

```text
student1 → Ajim, 30
student2 → Rahul, 25
```

The same class can produce objects containing different data.

## 9. Methods

A **method** is a function defined inside a class.

Methods represent the behavior/action of an object.

````python
class Student:

    def study(self):
        print("Student is studying")

student1 = Student()

student1.study()
````

Here:

```python
study()
```

is a method of the `Student` class.

### Function vs Method

```text
Function → defined independently
Method   → defined inside a class
```

## 10. Understanding `self`

`self` refers to the **current object**.

This is one of the most important concepts in Python OOP.

Example:

````python
class Student:

    def show(self):
        print("Current object", self)

student1 = Student()
student2 = Student()

student1.show()
student2.show()
````

When we call:

```python
student1.show()
```

Python automatically passes `student1` as the first argument.

Conceptually:

```python
Student.show(student1)
```

Therefore:

```python
self
```

refers to `student1`.

Similarly:

```python
student2.show()
```

means `self` refers to `student2`.

### Important rule

For an instance method, `self` represents the current object.

## 11. Using `self` to Store Object Data

Now we can combine attributes and methods.

````python
class Student:

    def set_data(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


student1 = Student()

student1.set_data("Ajim", 30)
student1.display()
````

### Understand this carefully

Inside:

```python
self.name = name
```

- `self.name` → attribute belonging to the current object
- `name` → value received by the method

For:

```python
student1.set_data("Ajim", 30)
```

the result is conceptually:

```text
student1.name → "Ajim"
student1.age  → 30
```

## 12. The `__init__()` Constructor

`__init__()` is automatically called when an object is created.

It is commonly called the **constructor** in Python learning materials.

Example:

````python
class Student:

    def __init__(self):

        print("Student object created")

student1 = Student()
student2 = Student()
````

When these objects are created:

```python
student1 = Student()
student2 = Student()
```

Python automatically calls `__init__()` for each object.

## 13. Constructor with Parameters

A much more useful pattern is to pass initial data when creating the object.

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

student1 = Student("Ajim", 25)

print(student1.name)
print(student1.age)
````

Now object creation and initialization happen together:

```python
student1 = Student("Ajim", 30)
```

The values are stored as:

```text
student1.name → "Ajim"
student1.age  → 30
```

## 14. Constructor + Method

This is the basic pattern you will use frequently in OOP.

````python
class Student:

    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Marks:", self.marks)

    def result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")


student1 = Student("Ajim", 28, 85)

student1.display()
student1.result()
````

## 15. Creating Multiple Objects with `__init__()`

One class can create many objects with different data.

````python
class Student:

    def __init__(self, name, age, marks):
        self.name =  name
        self.age = age
        self.marks = marks

    def display(self):
        print(f"{self.name} | Age: {self.age} | Marks: {self.marks}")

student1 = Student("Azeem", 28, 86)
student2= Student("Aryan", 25, 82)
student3 = Student("Sayeed", 25, 78)

student1.display()
student2.display()
student3.display()
````

### Object data is independent

Changing `student1` does not automatically change `student2`.

````python
student1.marks = 95

print("Student 1:", student1.marks)
print("Student 2:", student2.marks)
print("Student 3:", student3.marks)
````

## 16. Modifying Object Data

Object attributes can be modified after creation.

Example:

````python
class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

employee = Employee("Azeem", 75000)

print("Before:", employee.salary)
employee.salary = 85000
print("After:", employee.salary)
````

Later, when we study **encapsulation and properties**, we will learn how to control such changes and validate data.

## 17. Simple Real-World Example — Car

Let's model a real-world `Car`.

### Data

- brand
- model
- year

### Behavior

- start()
- stop()
- display_details()

````python
class Car:

    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def start(self):
        print(f"{self.brand} {self.model} started.")

    def stop(self):
        print(f"{self.brand} {self.model} stopped.")

    def display_details(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Year:", self.year)


car1 = Car("Toyota", "Camry", 2024)

car1.display_details()
car1.start()
car1.stop()
````

## 18. Another Real-World Example — Bank Account

This example prepares us for the encapsulation topic in the next notebook.

````python
class BankAccount:

    def __init__(self, account_holder, balance):
        self.account_holder = account_holder
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdraw successful.")

        else:
            print("Insufficient balance.")

    def show_balance(self):
        print("Account Holder:", self.account_holder)
        print("Balance:", self.balance)

account = BankAccount("Azeem", 75000)

account.show_balance()

account.deposit(786)
account.show_balance()

account.withdraw(1000)
account.show_balance()
````
