# 2. Attributes

An **attribute** is data associated with a class or object.

The two important types are:

```text
Attributes
    |
    ├── Instance Variables
    │       ↓
    │   Object-specific
    │
    └── Class Variables
            ↓
        Class-level/shared
```

# 3. Instance Variables

An instance variable belongs to a particular object.

It is normally created using:

```python
self.variable_name
```

````python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age


student1 = Student("Ajim", 30)
student2 = Student("Rahul", 25)

print(student1.name, student1.age)
print(student2.name, student2.age)
````

Each object has its own instance data:

```text
student1 → name=Ajim,  age=30
student2 → name=Rahul, age=25
```

Changing one object's instance variable does not normally change another object's value.

````python
student1.age = 31

print("Student 1:", student1.age)
print("Student 2:", student2.age)
````

# 4. Class Variables

A class variable is defined directly inside the class and represents class-level data.

Example:

````python
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name


student1 = Student("Ajim")
student2 = Student("Rahul")

print(student1.college)
print(student2.college)
print(Student.college)
````

Here:

```python
college = "ABC College"
```

is a class variable.

Conceptually:

```text
          Student class
               |
          college = ABC
            /       \
           /         \
      student1     student2
        name          name
        Ajim          Rahul
```

# 5. Instance Variable vs Class Variable

| Instance Variable | Class Variable |
|---|---|
| Belongs to an object | Defined at class level |
| Usually created with `self` | Defined directly in class |
| Can differ between objects | Common class-level value |
| Example: `self.name` | Example: `college` |
| Used for object-specific data | Used for shared/class data |

# 6. Inspecting Object Attributes

`__dict__` is useful for seeing instance data stored on an object.

````python
class Student:

    college = "ABC College"

    def __init__(self, name, age):
        self.name = name
        self.age = age


student = Student("Ajim", 30)

print("Object dictionary:")
print(student.__dict__)

print("\nClass variable:")
print(Student.college)
````

`student.__dict__` shows the instance attributes stored directly on `student`.

The class variable belongs to the class.

# 7. Updating Class Data

If we update the class variable through the class, objects that haven't overridden it see the new value.

````python
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name


student1 = Student("Ajim")
student2 = Student("Rahul")

print(student1.college)
print(student2.college)

Student.college = "XYZ College"

print(student1.college)
print(student2.college)
````

### Important behavior

If you do:

```python
student1.college = "Private College"
```

you create an instance attribute named `college` for `student1`; you do not change the class variable.

````python
class Student:

    college = "ABC College"

    def __init__(self, name):
        self.name = name


student1 = Student("Ajim")
student2 = Student("Rahul")

student1.college = "Private College"

print("student1:", student1.college)
print("student2:", student2.college)
print("Class:", Student.college)

print("student1 dictionary:", student1.__dict__)
````

# 8. Instance Methods

An instance method works with a particular object and normally receives `self`.

Example:

````python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)


student = Student("Ajim", 85)
student.display()
````

Because the method receives `self`, it can access object-specific data:

```python
self.name
self.marks
```

# 9. Instance Methods with Arguments

Instance methods can also receive additional arguments.

````python
class Calculator:

    def __init__(self, number):
        self.number = number

    def add(self, value):
        return self.number + value

    def multiply(self, value):
        return self.number * value


calc = Calculator(10)

print(calc.add(5))
print(calc.multiply(5))
````

Here:
- `self.number` → data stored in the object
- `value` → argument supplied when calling the method

# 10. Three Main Types of Methods

```text
Methods
   |
   ├── Instance Method → self
   ├── Class Method    → cls
   └── Static Method   → no automatic self/cls
```

Let's learn each type.

# 11. Class Methods

A class method works with the class itself.

It uses:

```python
@classmethod
```

and receives:

```python
cls
```

````python
class Student:

    college = "ABC College"

    @classmethod
    def show_college(cls):
        print("College:", cls.college)


Student.show_college()
````

Here `cls` refers to the `Student` class.

Conceptually:

```python
Student.show_college()
```

passes the class to the method, so:

```python
cls.college
```

refers to:

```python
Student.college
```

# 12. Class Method Can Modify Class Data

````python
class Student:

    college = "ABC College"

    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college


print(Student.college)

Student.change_college("XYZ College")

print(Student.college)
````

This is useful when an operation should affect the class as a whole.

# 13. Static Methods

A static method does not automatically receive `self` or `cls`.

Use:

```python
@staticmethod
```

when the operation is related to the class but doesn't need object or class state.

````python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def is_even(number):
        return number % 2 == 0


print(Calculator.add(10, 20))
print(Calculator.is_even(8))
print(Calculator.is_even(7))
````

The methods above don't need object data or class data.

They are grouped under `Calculator` because they are logically related to calculation.

# 14. All Three Method Types Together

````python
class Student:

    college = "ABC College"

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Instance method
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

    # Class method
    @classmethod
    def change_college(cls, new_college):
        cls.college = new_college

    # Static method
    @staticmethod
    def is_pass(marks):
        return marks >= 40


student = Student("Ajim", 85)

student.display()

print("College:", Student.college)

Student.change_college("XYZ College")
print("New college:", Student.college)

print("Pass:", Student.is_pass(student.marks))
````

# 15. Complete Comparison

| Feature | Instance Method | Class Method | Static Method |
|---|---|---|---|
| Decorator | None | `@classmethod` | `@staticmethod` |
| Automatic first argument | `self` | `cls` | None |
| Main focus | Object | Class | Independent logic |
| Direct access to instance data | Yes | No | No |
| Direct access to class data | Yes | Yes | No |
| Typical call | `obj.method()` | `Class.method()` | `Class.method()` |

### Easy memory trick

```text
self → object
cls  → class
static → neither automatically
```

# 16. When Should You Use Each?

### Instance Method

Use it when the method needs a specific object's data.

```python
student.display()
```

### Class Method

Use it when the operation needs or changes class-level information.

```python
Student.change_college(...)
```

### Static Method

Use it for related utility logic that doesn't need object/class state.

```python
Calculator.is_even(10)
```

# 17. Real-World Example — Employee

````python
class Employee:

    company = "ABC Technologies"
    employee_count = 0

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
        Employee.employee_count += 1

    # Instance method
    def display(self):
        print(f"Name: {self.name}")
        print(f"Salary: {self.salary}")
        print(f"Company: {self.company}")

    # Class method
    @classmethod
    def change_company(cls, new_company):
        cls.company = new_company

    # Static method
    @staticmethod
    def is_valid_salary(salary):
        return salary >= 15000


employee1 = Employee("Ajim", 50000)
employee2 = Employee("Rahul", 40000)

employee1.display()
print()
employee2.display()

print("\nEmployees:", Employee.employee_count)

Employee.change_company("XYZ Technologies")

print("\nAfter company change:")
employee1.display()

print("\nValid salary:", Employee.is_valid_salary(50000))
print("Valid salary:", Employee.is_valid_salary(10000))
````

# 18. Class Method as an Alternative Constructor

A useful practical pattern is using a class method to create an object from another input format.

Suppose we receive:

```text
Ajim,30,85
```

````python
class Student:

    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    @classmethod
    def from_string(cls, data):
        name, age, marks = data.split(",")
        return cls(name, int(age), float(marks))

    def display(self):
        print(self.name, self.age, self.marks)


student = Student.from_string("Ajim,30,85")
student.display()
````

The normal constructor:

```python
Student("Ajim", 30, 85)
```

Alternative constructor:

```python
Student.from_string("Ajim,30,85")
```

This pattern is useful when objects can be created from different input formats.

# 19. Common Beginner Mistakes

### Mistake 1 — Forgetting `self`

Incorrect:

```python
class Student:
    def display():
        print("Hello")
```

Correct:

```python
class Student:
    def display(self):
        print("Hello")
```

### Mistake 2 — Using `self` in a class method

A class method receives `cls`, not `self`.

### Mistake 3 — Expecting a static method to automatically know the object

A static method receives no automatic object reference.

### Mistake 4 — Confusing class and instance variables

```python
self.name = name
```

is instance data.

```python
college = "ABC"
```

inside the class is class-level data.

# 25. Interview Questions

### Q1. What is an instance variable?

A variable associated with a particular object, normally created using `self`.

### Q2. What is a class variable?

A variable defined at class level and shared through the class unless an instance overrides it.

### Q3. What is the difference between `self` and `cls`?

```text
self → current object
cls  → current class
```

### Q4. What is a class method?

A method decorated with `@classmethod` that automatically receives the class as `cls`.

### Q5. What is a static method?

A method decorated with `@staticmethod` that does not automatically receive `self` or `cls`.

### Q6. Can a class method access instance variables?

Not directly, because it does not receive a particular instance as `self`.

### Q7. Can a static method access class variables?

Not automatically. It can explicitly use a class name if appropriate.

### Q8. What is an alternative constructor?

A class method that provides another way to create an object, often by parsing or converting input.
