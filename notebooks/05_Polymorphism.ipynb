# 1. What is Polymorphism?

**Polymorphism** means:

> One interface, many forms.

The same method, function, or operator can behave differently depending on the object or data.

Example:

```text
Animal
  |
  ├── Dog → sound() → Bark
  ├── Cat → sound() → Meow
  └── Cow → sound() → Moo
```

The interface can remain:

```python
animal.sound()
```

while the behavior changes.

# 2. Real-Life Example

Think about the action `pay()`.

```text
Credit Card → pay() → card payment
UPI         → pay() → UPI payment
Cash        → pay() → cash payment
```

The action is the same, but its implementation depends on the object.

That is the basic idea of polymorphism.

# 3. Polymorphism with Functions

Python functions can work with many different types.

For example, `len()` works with strings, lists, tuples, and dictionaries.

````python
print(len("Python"))
print(len([10, 20, 30, 40]))
print(len((1, 2, 3)))
print(len({"name": "Ajim", "age": 30}))
````

Same function:

```python
len()
```

Different objects produce different behavior.

This is polymorphism.

# 4. Polymorphism with Inheritance

A common form of polymorphism happens when child classes override a parent method.

````python
class Animal:

    def sound(self):
        print("Animal makes a sound")


class Dog(Animal):

    def sound(self):
        print("Dog barks")


class Cat(Animal):

    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
````

Both objects use:

```python
sound()
```

but behave differently.

This is called **method overriding**.

# 5. Method Overriding

Method overriding occurs when a child class provides its own implementation of a method inherited from the parent.

````python
class Parent:

    def show(self):
        print("Parent show method")


class Child(Parent):

    def show(self):
        print("Child show method")


parent = Parent()
child = Child()

parent.show()
child.show()
````

Overriding requires an inheritance relationship:

```text
Parent
  ↓
Child
```

The child provides a different implementation of the same method.

# 6. Polymorphism Through a Common Interface

We can create one function that works with different objects.

````python
class Dog:

    def sound(self):
        return "Bark"


class Cat:

    def sound(self):
        return "Meow"


class Cow:

    def sound(self):
        return "Moo"


def make_sound(animal):
    print(animal.sound())


make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
````

The function does not need to know the exact class.

It only expects the object to provide:

```python
sound()
```

This is a powerful polymorphic design.

# 7. Duck Typing

Python often follows the idea:

> If it behaves like a duck, treat it like a duck.

Python can focus on what an object **can do** rather than its exact type.

````python
class Dog:

    def speak(self):
        print("Bark")


class Robot:

    def speak(self):
        print("Hello, I am a robot")


class Person:

    def speak(self):
        print("Hello")


def make_speak(obj):
    obj.speak()


make_speak(Dog())
make_speak(Robot())
make_speak(Person())
````

These classes do not need to inherit from the same parent.

They simply provide:

```python
speak()
```

That is **duck typing**.

# 8. Inheritance Polymorphism vs Duck Typing

| Inheritance Polymorphism | Duck Typing |
|---|---|
| Usually uses a common parent | Common parent is not required |
| Relationship is explicit | Behavior is what matters |
| Method overriding is common | Same required method is enough |
| More structured | More flexible |

Example:

```text
Inheritance:
Animal → Dog, Cat

Duck typing:
Dog, Robot, Person
       ↓
   all have speak()
```

# 9. Polymorphism with a List of Objects

````python
class Dog:

    def sound(self):
        print("Bark")


class Cat:

    def sound(self):
        print("Meow")


class Cow:

    def sound(self):
        print("Moo")


animals = [Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
````

The loop does not need to check the exact type.

It simply calls:

```python
animal.sound()
```

Each object supplies its own implementation.

# 11. Operator Overloading

Python allows custom classes to define operator behavior.

This is called **operator overloading**.

Common special methods include:

```python
__add__()
__sub__()
__mul__()
__eq__()
```

# 12. `__add__()` — Overloading `+`

````python
class Number:

    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


num1 = Number(10)
num2 = Number(20)

result = num1 + num2

print(result.value)
````

When Python evaluates:

```python
num1 + num2
```

it can use:

```python
num1.__add__(num2)
```

We have therefore defined what `+` means for our class.

# 13. `__str__()` — Readable Objects

````python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Student: {self.name}, Marks: {self.marks}"


student = Student("Ajim", 90)

print(student)
````

`__str__()` provides a user-friendly string representation of an object.

It is useful for:

```python
print(object)
str(object)
```

# 14. `__len__()` — Custom `len()`

````python
class Team:

    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)


team = Team(["Ajim", "Rahul", "Priya", "Amit"])

print(len(team))
````

Because `__len__()` is defined:

```python
len(team)
```

knows how to calculate the size of the custom object.

# 15. `__eq__()` — Custom Equality

````python
class Student:

    def __init__(self, name, roll_no):
        self.name = name
        self.roll_no = roll_no

    def __eq__(self, other):
        return self.roll_no == other.roll_no


student1 = Student("Ajim", 101)
student2 = Student("Rahul", 101)
student3 = Student("Priya", 102)

print(student1 == student2)
print(student1 == student3)
````

Here two students are considered equal when their `roll_no` is equal.

This demonstrates customized behavior for:

```python
==
```

# 16. Common Dunder Methods

| Method | Purpose |
|---|---|
| `__str__()` | String representation |
| `__len__()` | `len(obj)` |
| `__add__()` | `obj1 + obj2` |
| `__sub__()` | `obj1 - obj2` |
| `__mul__()` | `obj1 * obj2` |
| `__eq__()` | `obj1 == obj2` |
| `__lt__()` | `obj1 < obj2` |
| `__gt__()` | `obj1 > obj2` |
| `__contains__()` | `x in obj` |

They are commonly called **dunder methods** because their names have double underscores.

# 17. Complete Operator Overloading Example

Let's create a `Point` class.

````python
class Point:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __sub__(self, other):
        return Point(self.x - other.x, self.y - other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __str__(self):
        return f"({self.x}, {self.y})"


p1 = Point(2, 3)
p2 = Point(4, 5)

print("p1:", p1)
print("p2:", p2)

print("Addition:", p1 + p2)
print("Subtraction:", p1 - p2)

p3 = Point(2, 3)
print("p1 == p3:", p1 == p3)
````

# 18. Abstract Classes + Polymorphism

Python provides abstract classes through:

```python
from abc import ABC, abstractmethod
```

An abstract class can define a common interface that child classes must implement.

````python
from abc import ABC, abstractmethod


class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


class Dog(Animal):

    def sound(self):
        return "Bark"


class Cat(Animal):

    def sound(self):
        return "Meow"


dog = Dog()
cat = Cat()

print(dog.sound())
print(cat.sound())
````

The parent defines the required interface:

```python
sound()
```

Each child implements it differently.

This gives a structured form of polymorphism.

# 19. Real-World Example — Payment System

````python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCard(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Credit Card")


class UPI(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using UPI")


class BankTransfer(Payment):

    def pay(self, amount):
        print(f"Paid ₹{amount} using Bank Transfer")


payments = [
    CreditCard(),
    UPI(),
    BankTransfer()
]

for payment in payments:
    payment.pay(1000)
````

The application only needs to know:

```python
payment.pay(amount)
```

The actual implementation depends on the object.

This is one of the most useful patterns for real applications.

# 20. Polymorphism in Data Science / Machine Learning

Polymorphism is useful when different model classes expose the same interface.

For example:

```text
Model
 |
 ├── LinearModel
 ├── TreeModel
 └── NeuralNetworkModel

Common operations:
fit()
predict()
evaluate()
```

Application code can work with different models through the same interface.

````python
class LinearModel:

    def fit(self, X, y):
        print("Training Linear Model")

    def predict(self, X):
        return ["Linear prediction"]


class TreeModel:

    def fit(self, X, y):
        print("Training Tree Model")

    def predict(self, X):
        return ["Tree prediction"]


models = [LinearModel(), TreeModel()]

for model in models:
    model.fit([1, 2, 3], [10, 20, 30])
    print(model.predict([4]))
````

The application uses:

```python
model.fit()
model.predict()
```

without needing separate code for every model type.

# 21. Polymorphism vs Method Overriding

These concepts are closely related but not identical.

### Method overriding

A child class changes a method inherited from a parent.

```text
Parent
  ↓
Child
  ↓
same method name
different implementation
```

### Polymorphism

A broader concept where one interface can represent many behaviors.

Therefore:

```text
Method overriding → one technique for achieving polymorphism
Polymorphism       → broader OOP concept
```

# 22. Polymorphism vs Method Overloading

This is an important interview topic.

### Overriding

Occurs across inheritance:

```python
class Parent:
    def show(self):
        pass


class Child(Parent):
    def show(self):
        pass
```

### Traditional overloading

Some languages allow:

```text
add(a, b)
add(a, b, c)
add(a, b, c, d)
```

as separate methods.

Python does not support traditional signature-based method overloading in the same way.

A later method definition replaces the earlier one.

````python
class Calculator:

    def add(self, a, b):
        return a + b

    # This replaces the previous add() definition
    def add(self, a, b, c):
        return a + b + c


calc = Calculator()

print(calc.add(10, 20, 30))
````

Python commonly uses alternatives such as:

- Default arguments
- `*args`
- `**kwargs`

Example:

````python
class Calculator:

    def add(self, *numbers):
        return sum(numbers)


calc = Calculator()

print(calc.add(10, 20))
print(calc.add(10, 20, 30))
print(calc.add(1, 2, 3, 4, 5))
````

# 23. Practical Example — Notification System

````python
class EmailNotification:

    def send(self, message):
        print("Email:", message)


class SMSNotification:

    def send(self, message):
        print("SMS:", message)


class PushNotification:

    def send(self, message):
        print("Push Notification:", message)


def notify(notification, message):
    notification.send(message)


notifications = [
    EmailNotification(),
    SMSNotification(),
    PushNotification()
]

for notification in notifications:
    notify(notification, "Your order has been shipped.")
````

The `notify()` function does not care about the exact notification class.

It only expects:

```python
send()
```

This is polymorphism through a common behavior.

# 24. Practical Example — Shapes

Different shapes can implement the same `area()` method.

````python
class Circle:

    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2


class Rectangle:

    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


class Triangle:

    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height


shapes = [
    Circle(5),
    Rectangle(10, 4),
    Triangle(8, 6)
]

for shape in shapes:
    print("Area:", shape.area())
````

Every object supports:

```python
area()
```

but each class calculates it differently.

That is polymorphism.

# 25. Practical Example — Employees

Different employee types can calculate salary differently.

````python
class FullTimeEmployee:

    def calculate_salary(self):
        return 50000


class PartTimeEmployee:

    def calculate_salary(self):
        return 25000


class Intern:

    def calculate_salary(self):
        return 10000


employees = [
    FullTimeEmployee(),
    PartTimeEmployee(),
    Intern()
]

for employee in employees:
    print("Salary:", employee.calculate_salary())
````

The interface remains:

```python
employee.calculate_salary()
```

while the implementation changes for each employee type.

# 26. Common Beginner Mistakes

### Mistake 1 — Thinking polymorphism means only overriding

Method overriding is one form of polymorphism, not the complete definition.

### Mistake 2 — Confusing overriding and overloading

```text
Overriding → child changes inherited behavior
Overloading → same name with different signatures
```

### Mistake 3 — Using type checks everywhere

Polymorphic design can often remove large chains of:

```python
if isinstance(...)
```

### Mistake 4 — Overusing operator overloading

Operators should have intuitive meanings.

### Mistake 5 — Forgetting the common interface

Objects used polymorphically need compatible behavior, such as `area()`, `pay()`, or `predict()`.

# 33. Interview Questions

### Q1. What is polymorphism?

Polymorphism allows a common interface to represent different implementations or behaviors.

### Q2. Give a real-world example.

Different payment methods can provide the same `pay()` operation while implementing it differently.

### Q3. What is method overriding?

When a child class provides its own implementation of a method inherited from a parent class.

### Q4. What is duck typing?

A Python approach where an object's usable behavior matters more than its exact type.

### Q5. What is operator overloading?

Defining how operators such as `+`, `-`, `==`, or `<` behave for custom objects using special methods.

### Q6. What are dunder methods?

Special methods whose names start and end with double underscores, such as `__init__`, `__str__`, and `__add__`.

### Q7. Does Python support traditional method overloading?

Python does not support traditional signature-based method overloading in the same way as languages such as Java. Default arguments and `*args` are common alternatives.

### Q8. What is the difference between overriding and overloading?

```text
Overriding:
Parent → Child
Same method → different implementation

Overloading:
Same method name → different parameter signatures
```

### Q9. Why is polymorphism useful?

It reduces type-specific code, improves flexibility, supports extensibility, and allows different implementations to be used through a common interface.

### Q10. Why is polymorphism useful in ML projects?

Different models can expose common operations such as:

```python
fit()
predict()
evaluate()
```

so application code can work with multiple model implementations.
