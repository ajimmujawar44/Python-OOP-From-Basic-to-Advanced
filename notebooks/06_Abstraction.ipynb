## 1. What is Abstraction?

**Abstraction means hiding unnecessary implementation details and exposing only the essential functionality.**

Think about driving a car: you use the steering wheel, accelerator and brake without needing to know how the engine internally works.

```text
User
  │
  ▼
Simple Interface
  │
  ▼
Complex Internal Implementation
```

The user focuses on **what an object does**, while the class handles **how it does it**.

## 2. Why is Abstraction Important?

Abstraction helps us:
1. Reduce complexity
2. Hide implementation details
3. Define a common interface
4. Improve maintainability
5. Make large systems easier to extend
6. Encourage consistent design

A good abstraction exposes only what the caller needs.

## 3. Real-World Example

A payment system may expose:

```python
payment.pay(1000)
```

The caller does not need to know whether the payment is processed using a card, UPI, bank transfer, or another internal mechanism.

The implementation can vary while the public operation remains consistent.

## 4. Abstraction in Python

Python supports abstraction through the `abc` module:

```python
from abc import ABC, abstractmethod
```

- `ABC` → base class for abstract classes
- `@abstractmethod` → declares a method that subclasses must implement

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
````

`Animal` is now an **abstract class** because it contains an abstract method.

You normally do not create an object directly from an abstract class when it still has unimplemented abstract methods.

````python
# This raises TypeError:
# animal = Animal()
````

## 5. Abstract Method

An abstract method defines a required operation without providing the complete implementation in the base class.

```python
@abstractmethod
def sound(self):
    pass
```

Any concrete subclass must implement `sound()`.

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        return "Bark"

class Cat(Animal):
    def sound(self):
        return "Meow"

dog = Dog()
cat = Cat()

print(dog.sound())
print(cat.sound())
````

## 6. Abstract Class with Multiple Methods

````python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

rectangle = Rectangle(10, 5)
print("Area:", rectangle.area())
print("Perimeter:", rectangle.perimeter())
````

## 7. Concrete Class

A **concrete class** provides implementations for all required abstract methods and can be instantiated.

```text
Abstract Class
      │
      ├── Abstract methods
      │
      ▼
Concrete Subclass
      │
      ├── Implement methods
      ▼
Object can be created
```

## 8. Abstract Class Can Contain Concrete Methods

An abstract class does not have to contain only abstract methods. It can contain normal methods as well.

````python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"Employee: {self.name}"

    @abstractmethod
    def calculate_salary(self):
        pass

class Developer(Employee):
    def calculate_salary(self):
        return 80000

class Manager(Employee):
    def calculate_salary(self):
        return 120000

for employee in [Developer("Amit"), Manager("Neha")]:
    print(employee.introduce())
    print("Salary:", employee.calculate_salary())
````

## 9. Abstraction Using a Payment System

This is a common practical design: every payment provider must implement `pay()`.

````python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

class CreditCardPayment(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using Credit Card"

class UPIPayment(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using UPI"

class BankTransferPayment(Payment):
    def pay(self, amount):
        return f"Paid ₹{amount} using Bank Transfer"

for method in [CreditCardPayment(), UPIPayment(), BankTransferPayment()]:
    print(method.pay(1500))
````

## 10. Abstraction + Polymorphism

Abstraction often works together with polymorphism. The application can depend on the abstract interface rather than a specific implementation.

````python
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass

class EmailNotification(Notification):
    def send(self, message):
        print("Email:", message)

class SMSNotification(Notification):
    def send(self, message):
        print("SMS:", message)

def notify(service, message):
    service.send(message)

notify(EmailNotification(), "Welcome!")
notify(SMSNotification(), "Your OTP is 1234")
````

## 11. Abstraction vs Encapsulation

| Abstraction | Encapsulation |
|---|---|
| Hides unnecessary implementation details | Controls access to internal state |
| Focuses on **what** an object does | Focuses on **how data is protected/controlled** |
| Uses abstract classes/interfaces and clean APIs | Uses classes, properties, methods, naming conventions |
| Reduces conceptual complexity | Protects object state and maintains valid state |

### Easy memory trick
**Abstraction = What**

**Encapsulation = How data is controlled**

## 12. Abstraction vs Inheritance

Inheritance is a mechanism for deriving a new class from an existing class.

Abstraction is a design concept for exposing essential behavior and hiding implementation details.

They are often used together: an abstract base class can define the contract and subclasses can inherit from it.

## 13. Abstract Properties

Properties can also be abstract.

````python
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @property
    @abstractmethod
    def fuel_type(self):
        pass

class ElectricCar(Vehicle):
    @property
    def fuel_type(self):
        return "Electricity"

class PetrolCar(Vehicle):
    @property
    def fuel_type(self):
        return "Petrol"

print(ElectricCar().fuel_type)
print(PetrolCar().fuel_type)
````

## 14. Abstract Class with Constructor

An abstract class can define shared initialization logic.

````python
from abc import ABC, abstractmethod

class Model(ABC):
    def __init__(self, name):
        self.name = name

    def info(self):
        return f"Model: {self.name}"

    @abstractmethod
    def predict(self, data):
        pass

class LinearModel(Model):
    def predict(self, data):
        return [1 for _ in data]

model = LinearModel("Linear Regression")
print(model.info())
print(model.predict([10, 20, 30]))
````

## 15. Data Science Example — ML Model Interface

Abstraction is useful when multiple ML models should provide the same operations.

````python
from abc import ABC, abstractmethod

class MLModel(ABC):
    @abstractmethod
    def train(self, X, y):
        pass

    @abstractmethod
    def predict(self, X):
        pass

class SimpleClassifier(MLModel):
    def train(self, X, y):
        print(f"Training on {len(X)} samples")

    def predict(self, X):
        return [0] * len(X)

model = SimpleClassifier()
model.train([[1], [2], [3]], [0, 1, 0])
print(model.predict([[4], [5]]))
````

## 16. Data Science Example — Data Preprocessor

Different preprocessors can follow the same contract.

````python
from abc import ABC, abstractmethod

class Preprocessor(ABC):
    @abstractmethod
    def fit(self, data):
        pass

    @abstractmethod
    def transform(self, data):
        pass

class StandardPreprocessor(Preprocessor):
    def fit(self, data):
        self.mean = sum(data) / len(data)
        return self

    def transform(self, data):
        return [x - self.mean for x in data]

processor = StandardPreprocessor()
processor.fit([10, 20, 30])
print(processor.transform([10, 20, 30]))
````

## 17. Interface-Like Design in Python

Python does not have a separate `interface` keyword like Java. Abstract base classes can provide an interface-like contract.

For example, if every storage class must implement `save()`, the abstract base class defines that requirement.

````python
from abc import ABC, abstractmethod

class Storage(ABC):
    @abstractmethod
    def save(self, data):
        pass

class FileStorage(Storage):
    def save(self, data):
        return f"Saving {data} to a file"

class DatabaseStorage(Storage):
    def save(self, data):
        return f"Saving {data} to a database"

for storage in [FileStorage(), DatabaseStorage()]:
    print(storage.save("record"))
````

## 18. What Happens If a Subclass Does Not Implement an Abstract Method?

````python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    pass

# This raises TypeError because Dog does not implement sound():
# dog = Dog()
````

## 19. Abstract Method Can Have an Implementation

An abstract method can contain code. A subclass still has to override it before it can be instantiated. This can be useful when subclasses should extend shared behavior.

````python
from abc import ABC, abstractmethod

class Report(ABC):
    @abstractmethod
    def generate(self):
        print("Starting report generation")

class SalesReport(Report):
    def generate(self):
        super().generate()
        print("Generating sales report")

report = SalesReport()
report.generate()
````

## 20. Practical Design Pattern

A useful abstraction design is:

```text
                Abstract Service
                       │
             defines required methods
                       │
        ┌──────────────┼──────────────┐
        ▼              ▼              ▼
   Implementation A  Implementation B  C
        │              │              │
        └──────────────┼──────────────┘
                       ▼
                 Application
```

The application depends on the common contract, not the internal implementation.

## 21. Benefits of Abstraction

### 1. Simplicity
Users interact with a smaller interface.

### 2. Maintainability
Internal implementations can change without changing callers.

### 3. Extensibility
New subclasses can be added while following the same contract.

### 4. Consistency
Every implementation must provide required operations.

### 5. Reduced coupling
High-level code can depend on an abstraction instead of a concrete class.

## 22. When Should You Use Abstraction?

Use abstraction when:
- Multiple classes share a common contract.
- You want to enforce required methods.
- Implementations may vary.
- The internal implementation is complex.
- A project is large enough to benefit from clear boundaries.

For a tiny script, an abstract class may add unnecessary complexity.

## 23. Common Mistakes

### Mistake 1 — Instantiating an incomplete abstract class

```python
class A(ABC):
    @abstractmethod
    def run(self):
        pass

a = A()  # TypeError
```

### Mistake 2 — Forgetting to implement an abstract method
A subclass remains abstract until all inherited abstract methods are implemented.

### Mistake 3 — Confusing abstraction with encapsulation
They solve related but different design problems.

## 24. Complete Example — Ride Booking System

````python
from abc import ABC, abstractmethod

class Ride(ABC):
    def __init__(self, distance):
        if distance <= 0:
            raise ValueError("Distance must be positive")
        self.distance = distance

    @abstractmethod
    def calculate_fare(self):
        pass

    def booking_summary(self):
        return f"Distance: {self.distance} km | Fare: ₹{self.calculate_fare():.2f}"

class BikeRide(Ride):
    def calculate_fare(self):
        return 30 + self.distance * 10

class CarRide(Ride):
    def calculate_fare(self):
        return 50 + self.distance * 15

class PremiumRide(Ride):
    def calculate_fare(self):
        return 100 + self.distance * 25

for ride in [BikeRide(5), CarRide(5), PremiumRide(5)]:
    print(ride.booking_summary())
````

## 27. Interview Questions — Quick Answers

**Q1. What is abstraction?**

Hiding unnecessary implementation details and exposing essential functionality.

**Q2. What is an abstract class?**

A class intended to define a common contract and/or shared behavior; with unimplemented abstract methods, it cannot be instantiated.

**Q3. Which module provides abstraction support?**

The `abc` module.

**Q4. What is `ABC`?**

A base class used to define abstract base classes.

**Q5. What does `@abstractmethod` do?**

It marks a method as abstract and requires concrete subclasses to implement it before instantiation.

**Q6. Can an abstract class have a constructor?**

Yes.

**Q7. Can an abstract class have concrete methods?**

Yes.

**Q8. Can an abstract method contain code?**

Yes, although subclasses must still implement it.

**Q9. Does Python have an `interface` keyword?**

No separate keyword; ABCs can provide interface-like contracts.
