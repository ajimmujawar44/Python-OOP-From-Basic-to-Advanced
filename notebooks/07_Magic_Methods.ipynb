# 07 — Magic Methods in Python OOP

**Level:** Beginner → Advanced

Magic methods, also called **dunder methods**, are special methods whose names begin and end with double underscores, such as `__init__`, `__str__`, and `__add__`.

### Learning Objectives
- Understand magic/dunder methods
- Learn how Python invokes them
- Master object lifecycle, representation, comparison, arithmetic, containers, iteration, callables, and context managers
- Understand operator overloading
- Apply magic methods to Data Science and ML dataset/model designs


## 1. What Are Magic Methods?

Magic methods are special hooks used by Python's data model.

Examples:

```text
__init__      → initialize an object
__new__       → create an object
__str__       → user-friendly representation
__repr__      → developer-friendly representation
__len__       → len(obj)
__eq__        → obj == other
__add__       → obj + other
__getitem__   → obj[index]
__iter__      → iteration
__call__      → obj(...)
__enter__     → enter a with block
__exit__      → leave a with block
```

Python usually calls these methods automatically when you use normal Python syntax.


## 2. Why Are Magic Methods Important?

They make custom classes behave like built-in Python objects.

For example, you can make a class support:

```python
print(obj)
len(obj)
obj + other
obj == other
obj[0]
for x in obj:
    ...
obj(...)
```

This is a major part of writing **Pythonic OOP**.


````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

student = Student("Ajim", 30)
print(student.name)
print(student.age)
````

## 3. `__init__()` — Object Initialization

`__init__()` is called automatically after an instance has been created. It initializes the object's state.

> Advanced note: object creation itself is associated with `__new__()`, while `__init__()` initializes the returned instance.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __str__(self):
        return f"Student: {self.name}, Marks: {self.marks}"

student = Student("Ajim", 92)
print(student)
print(str(student))
````

## 4. `__str__()` — Friendly Representation

`__str__()` controls the human-friendly string returned by `str(obj)` and normally used by `print(obj)`.

It must return a string.


````python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def __repr__(self):
        return f"Student(name={self.name!r}, marks={self.marks!r})"

student = Student("Ajim", 92)
print(repr(student))
````

## 5. `__repr__()` — Developer Representation

`__repr__()` should give a useful, unambiguous representation for debugging and development.

### `__str__()` vs `__repr__()`

| Method | Purpose |
|---|---|
| `__str__()` | User-friendly output |
| `__repr__()` | Developer/debugging representation |


````python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

team = Team(["A", "B", "C", "D"])
print(len(team))
````

## 6. `__len__()` — Custom Length

`__len__()` defines what `len(obj)` means for your class.


````python
class ShoppingCart:
    def __init__(self, items):
        self.items = items

    def __bool__(self):
        return len(self.items) > 0

empty_cart = ShoppingCart([])
full_cart = ShoppingCart(["Laptop"])

print(bool(empty_cart))
print(bool(full_cart))

if full_cart:
    print("Cart has items")
````

## 7. `__bool__()` — Truth Value

`__bool__()` controls whether an object is considered truthy or falsy.

If `__bool__()` is absent, Python can use `__len__()` as a fallback.


````python
class Student:
    def __init__(self, roll_no, name):
        self.roll_no = roll_no
        self.name = name

    def __eq__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.roll_no == other.roll_no

s1 = Student(101, "Ajim")
s2 = Student(101, "Rahul")
s3 = Student(102, "Sara")

print(s1 == s2)
print(s1 == s3)
````

## 8. `__eq__()` — Equality

`__eq__()` defines the meaning of `==`.

Other comparison hooks include:

```text
__ne__  → !=
__lt__  → <
__le__  → <=
__gt__  → >
__ge__  → >=
```


````python
class Product:
    def __init__(self, price):
        self.price = price

    def __lt__(self, other):
        if not isinstance(other, Product):
            return NotImplemented
        return self.price < other.price

    def __gt__(self, other):
        if not isinstance(other, Product):
            return NotImplemented
        return self.price > other.price

p1 = Product(500)
p2 = Product(800)

print(p1 < p2)
print(p1 > p2)
````

## 9. `__add__()` — Operator Overloading

`__add__()` defines how `+` behaves for instances of your class.

This is called **operator overloading**.


````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if not isinstance(other, Point):
            return NotImplemented
        return Point(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

p1 = Point(2, 3)
p2 = Point(4, 5)

print(p1 + p2)
````

## 10. Common Arithmetic Magic Methods

| Operator | Method |
|---|---|
| `+` | `__add__` |
| `-` | `__sub__` |
| `*` | `__mul__` |
| `/` | `__truediv__` |
| `//` | `__floordiv__` |
| `%` | `__mod__` |
| `**` | `__pow__` |

There are also reflected methods such as `__radd__()` and in-place methods such as `__iadd__()`.


````python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __sub__(self, other):
        return Number(self.value - other.value)

    def __mul__(self, other):
        return Number(self.value * other.value)

    def __repr__(self):
        return f"Number({self.value})"

a = Number(10)
b = Number(5)

print(a + b)
print(a - b)
print(a * b)
````

## 11. `__getitem__()` — Indexing and Slicing

`__getitem__()` enables collection-like syntax:

```python
obj[0]
obj[1:4]
```


````python
class Dataset:
    def __init__(self, data):
        self.data = data

    def __getitem__(self, index):
        return self.data[index]

dataset = Dataset([10, 20, 30, 40, 50])

print(dataset[0])
print(dataset[1:4])
````

## 12. `__setitem__()` — Assignment

`__setitem__()` enables:

```python
obj[index] = value
```


````python
class Dataset:
    def __init__(self, data):
        self.data = data

    def __getitem__(self, index):
        return self.data[index]

    def __setitem__(self, index, value):
        self.data[index] = value

    def __repr__(self):
        return repr(self.data)

dataset = Dataset([10, 20, 30])
dataset[1] = 99
print(dataset)
````

## 13. `__contains__()` — Membership

`__contains__()` controls:

```python
value in object
```


````python
class Course:
    def __init__(self, topics):
        self.topics = topics

    def __contains__(self, topic):
        return topic.lower() in [t.lower() for t in self.topics]

course = Course(["Python", "Machine Learning", "Deep Learning"])

print("python" in course)
print("SQL" in course)
````

## 14. `__iter__()` — Making an Object Iterable

`__iter__()` lets your object work with a `for` loop.

For collection-like classes, returning the underlying collection's iterator is often enough.


````python
class Numbers:
    def __init__(self, values):
        self.values = values

    def __iter__(self):
        return iter(self.values)

numbers = Numbers([10, 20, 30])

for number in numbers:
    print(number)
````

## 15. `__next__()` — Custom Iterators

An iterator implements `__next__()` and raises `StopIteration` when there are no more values.


````python
class Counter:
    def __init__(self, start, end):
        self.current = start
        self.end = end

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.end:
            raise StopIteration
        value = self.current
        self.current += 1
        return value

for value in Counter(1, 5):
    print(value)
````

## 16. `__call__()` — Callable Objects

`__call__()` allows an instance to be used like a function:

```python
obj(...)
```

This is useful for function-like objects, decorators, processors, and ML components.


````python
class Greeter:
    def __init__(self, greeting):
        self.greeting = greeting

    def __call__(self, name):
        return f"{self.greeting}, {name}!"

greeter = Greeter("Hello")
print(greeter("Ajim"))
````

## 17. `__enter__()` and `__exit__()` — Context Managers

These methods support the `with` statement.

- `__enter__()` runs when entering the block.
- `__exit__()` runs when leaving the block.

They are commonly used for reliable resource management.


````python
class Resource:
    def __enter__(self):
        print("Resource opened")
        return self

    def work(self):
        print("Working with resource")

    def __exit__(self, exc_type, exc_value, traceback):
        print("Resource closed")

with Resource() as resource:
    resource.work()
````

## 18. `__del__()` — Finalization

`__del__()` may run when an object is being finalized.

**Do not rely on it for critical cleanup.** Finalization timing can be unpredictable. Prefer context managers (`with`) or explicit cleanup for files, connections, locks, and other resources.


````python
class Demo:
    def __init__(self, name):
        self.name = name

    def __del__(self):
        print(f"Finalizing {self.name}")

obj = Demo("Example")
del obj
````

## 19. `__new__()` — Advanced Object Creation

`__new__()` is responsible for creating and returning a new instance. It runs before `__init__()`.

Most normal classes do not need to override it.


````python
class Demo:
    def __new__(cls, *args, **kwargs):
        print("__new__ called")
        instance = super().__new__(cls)
        return instance

    def __init__(self, value):
        print("__init__ called")
        self.value = value

obj = Demo(100)
print(obj.value)
````

## 20. Object Lifecycle

A simplified lifecycle is:

```text
Class called
    ↓
__new__()
    ↓
Object created
    ↓
__init__()
    ↓
Object used
    ↓
Object eventually finalized
```

`__new__()` creates/returns the instance; `__init__()` initializes it.


## 21. `__hash__()` — Hashable Objects

`__hash__()` supplies a hash value for objects that can be used in sets or as dictionary keys.

If you customize value-based `__eq__()`, think carefully about hashing. Mutable objects generally should not be hashable based on mutable state.


````python
class Student:
    def __init__(self, roll_no):
        self.roll_no = roll_no

    def __eq__(self, other):
        if not isinstance(other, Student):
            return NotImplemented
        return self.roll_no == other.roll_no

s1 = Student(101)
s2 = Student(101)

print(s1 == s2)
print(Student.__hash__)  # None: Python makes the class unhashable here
````

## 22. Value Object with `__hash__()`

For an immutable-style value object, equality and hashing can be defined consistently.


````python
class Coordinate:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        if not isinstance(other, Coordinate):
            return NotImplemented
        return (self.x, self.y) == (other.x, other.y)

    def __hash__(self):
        return hash((self.x, self.y))

    def __repr__(self):
        return f"Coordinate({self.x}, {self.y})"

c1 = Coordinate(10, 20)
c2 = Coordinate(10, 20)

print(c1 == c2)
print(c2 in {c1})
````

## 23. `__format__()` — Custom Formatting

`__format__()` controls:

```python
format(obj, "spec")
```

It can provide domain-specific formatting.


````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __format__(self, spec):
        if spec == "currency":
            return f"₹{self.amount:,.2f}"
        return format(self.amount, spec)

money = Money(125000)
print(format(money, "currency"))
````

## 24. `__dict__` and Object State

`__dict__` is not a magic method, but it is useful when learning Python objects. For many normal classes it contains instance attributes.


````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

employee = Employee("Ajim", 50000)
print(employee.__dict__)
````

## 25. `__slots__` — Advanced Class Feature

`__slots__` is not a magic method. It can restrict instance attributes and may reduce per-instance memory usage.

Use it deliberately because it changes normal instance behavior.


````python
class Point:
    __slots__ = ("x", "y")

    def __init__(self, x, y):
        self.x = x
        self.y = y

point = Point(10, 20)
print(point.x, point.y)
````

## 26. Practical Example — Vector Class

Combine representation, length, addition, and equality.


````python
class Vector:
    def __init__(self, values):
        self.values = list(values)

    def __repr__(self):
        return f"Vector({self.values})"

    def __len__(self):
        return len(self.values)

    def __add__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if len(self) != len(other):
            raise ValueError("Vectors must have the same length")
        return Vector(a + b for a, b in zip(self.values, other.values))

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        return self.values == other.values

v1 = Vector([1, 2, 3])
v2 = Vector([4, 5, 6])

print(v1)
print(len(v1))
print(v1 + v2)
print(v1 == Vector([1, 2, 3]))
````

## 27. Practical Example — Custom Dataset

Magic methods make a custom dataset feel like a normal Python collection.

Common ML dataset behaviors include:

```python
len(dataset)
dataset[index]
for sample in dataset:
    ...
```


````python
class MLDataset:
    def __init__(self, X, y):
        self.X = X
        self.y = y

    def __len__(self):
        return len(self.X)

    def __getitem__(self, index):
        return self.X[index], self.y[index]

    def __iter__(self):
        return (self[i] for i in range(len(self)))

dataset = MLDataset(
    [[1, 2], [3, 4], [5, 6]],
    [0, 1, 0]
)

print("Dataset size:", len(dataset))
print("First sample:", dataset[0])

for sample in dataset:
    print(sample)
````

## 29. Magic Methods and Polymorphism

Different classes can implement the same special method differently.

For example, `+` can mean vector addition, money addition, or another domain-specific operation depending on the operand types.

This is closely related to Python's polymorphic behavior.


````python
class Price:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Price(self.value + other.value)

    def __repr__(self):
        return f"₹{self.value}"

class Distance:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Distance(self.value + other.value)

    def __repr__(self):
        return f"{self.value} km"

print(Price(100) + Price(50))
print(Distance(10) + Distance(5))
````

## 30. Magic Methods vs Normal Methods

### Normal method

```python
obj.calculate()
```

You explicitly call it.

### Magic method

```python
obj + other
len(obj)
print(obj)
```

Python's data model invokes the appropriate special method.

The special method is still ordinary Python code; it simply has a defined protocol.

## 37. Advanced Challenge — ML Model Wrapper

Create an abstract ML model interface.

Required methods:

```python
model.train(X, y)
model.predict(X)
model.evaluate(X, y)
```

Add:

- `__repr__()` for model information
- `__call__()` so `model(X)` performs prediction
- `__bool__()` to indicate whether training has happened

This is a useful design pattern for model wrappers.


````python
from abc import ABC, abstractmethod

class MLModel(ABC):
    @abstractmethod
    def train(self, X, y):
        pass

    @abstractmethod
    def predict(self, X):
        pass

    @abstractmethod
    def evaluate(self, X, y):
        pass

    def __call__(self, X):
        return self.predict(X)


class SimpleModel(MLModel):
    def __init__(self):
        self.is_trained = False

    def train(self, X, y):
        self.is_trained = True

    def predict(self, X):
        if not self.is_trained:
            raise RuntimeError("Train the model first")
        return [0 for _ in X]

    def evaluate(self, X, y):
        predictions = self.predict(X)
        return sum(p == actual for p, actual in zip(predictions, y)) / len(y)

    def __repr__(self):
        return f"SimpleModel(is_trained={self.is_trained})"

    def __bool__(self):
        return self.is_trained

model = SimpleModel()

print(model)
print(bool(model))

model.train([[1], [2]], [0, 1])

print(model)
print(bool(model))
print(model([[3], [4]]))
````

## 40. Final Mental Model

Think of magic methods as **hooks into Python's data model**:

```text
Your object
   │
   ├── print(obj) ──────────→ __str__()
   ├── repr(obj) ───────────→ __repr__()
   ├── len(obj) ────────────→ __len__()
   ├── bool(obj) ───────────→ __bool__()
   ├── obj + other ─────────→ __add__()
   ├── obj == other ────────→ __eq__()
   ├── obj < other ─────────→ __lt__()
   ├── obj[index] ──────────→ __getitem__()
   ├── value in obj ────────→ __contains__()
   ├── for x in obj ────────→ __iter__()
   ├── obj(...) ────────────→ __call__()
   └── with obj: ───────────→ __enter__ / __exit__
```

### Key Idea

> **Python operation → special method → your class defines the behavior.**
