## 1. Three Common Method Types

| Type | First parameter | Works mainly with | Decorator |
|---|---|---|---|
| Instance method | `self` | Object state | None |
| Class method | `cls` | Class state | `@classmethod` |
| Static method | None | Utility logic | `@staticmethod` |

````python
class Student:
    school = "Data Science Academy"

    def __init__(self, name):
        self.name = name

    def introduce(self):
        return f"I am {self.name}"

    @classmethod
    def school_name(cls):
        return cls.school

    @staticmethod
    def is_adult(age):
        return age >= 18

s = Student("Ajim")
print(s.introduce())
print(Student.school_name())
print(Student.is_adult(30))
````

## 2. Instance Methods

Instance methods receive `self` and can read or modify individual object attributes.

````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def increase_salary(self, percentage):
        self.salary += self.salary * percentage / 100

e = Employee("Ajim", 50000)
e.increase_salary(10)
print(e.salary)
````

## 3. Class Methods

A class method receives `cls`, referring to the class itself. It is useful when behavior needs class-level state or should create objects through the class.

````python
class Employee:
    company = "Tech Corp"

    def __init__(self, name):
        self.name = name

    @classmethod
    def company_name(cls):
        return cls.company

print(Employee.company_name())
````

## 4. Class Method as an Alternative Constructor

One of the most useful applications of `@classmethod` is a **factory/alternative constructor**.

````python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    @classmethod
    def from_string(cls, text):
        name, age = text.split(",")
        return cls(name.strip(), int(age))

p = Person.from_string("Ajim, 30")
print(p.name, p.age)
````

## 5. Why `cls` Matters

Using `cls(...)` instead of hard-coding the class name makes a class method friendly to inheritance: subclasses can inherit the factory and receive subclass instances.

````python
class Person:
    def __init__(self, name):
        self.name = name

    @classmethod
    def from_name(cls, name):
        return cls(name)

class Student(Person):
    pass

student = Student.from_name("Ajim")
print(type(student).__name__)
````

## 6. Static Methods

A static method does not receive `self` or `cls`. It is placed inside a class because the function is logically related to that class.

````python
class MathTools:
    @staticmethod
    def add(a, b):
        return a + b

    @staticmethod
    def is_even(number):
        return number % 2 == 0

print(MathTools.add(10, 20))
print(MathTools.is_even(8))
````

## 7. When Should You Use a Static Method?

Use it when:
- the logic belongs conceptually to the class
- it does not need object state
- it does not need class state

If it is completely unrelated to the class, a normal module-level function may be clearer.

## 8. Class Method vs Static Method

| Feature | Class Method | Static Method |
|---|---|---|
| Decorator | `@classmethod` | `@staticmethod` |
| First argument | `cls` | None |
| Access class state | Yes | No automatically |
| Alternative constructor | Common | Not normally |
| Utility function | Sometimes | Common |

## 9. `@property` — Pythonic Getters

`@property` lets a method be accessed like an attribute. This is useful when reading a value requires logic or validation.

````python
class Circle:
    def __init__(self, radius):
        self._radius = radius

    @property
    def radius(self):
        return self._radius

c = Circle(5)
print(c.radius)
````

## 10. Property Setter

A setter lets you control assignment while keeping natural attribute syntax.

````python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    @property
    def radius(self):
        return self._radius

    @radius.setter
    def radius(self, value):
        if value <= 0:
            raise ValueError("Radius must be positive")
        self._radius = value

c = Circle(5)
print(c.radius)
c.radius = 10
print(c.radius)
````

## 11. Property Deleter

`@property_name.deleter` can define what happens when `del obj.property` is used. Use it only when deletion has a meaningful domain behavior.

````python
class Account:
    def __init__(self, owner):
        self.owner = owner

    @property
    def owner(self):
        return self._owner

    @owner.deleter
    def owner(self):
        self._owner = None

    @owner.setter
    def owner(self, value):
        self._owner = value

a = Account("Ajim")
print(a.owner)
del a.owner
print(a.owner)
````

## 12. Read-Only Property

If you define a property without a setter, users cannot assign to it through that property.

````python
class Employee:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    @property
    def full_name(self):
        return f"{self.first_name} {self.last_name}"

e = Employee("Ajim", "Mujawar")
print(e.full_name)
````

## 13. Property for Computed Values

Properties are especially useful for values derived from other attributes.

````python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height

r = Rectangle(10, 5)
print(r.area)
````

## 14. Validation with Properties

Properties can keep an object in a valid state by validating data before storing it.

````python
class Product:
    def __init__(self, price):
        self.price = price

    @property
    def price(self):
        return self._price

    @price.setter
    def price(self, value):
        if not isinstance(value, (int, float)):
            raise TypeError("Price must be numeric")
        if value < 0:
            raise ValueError("Price cannot be negative")
        self._price = value

p = Product(999)
print(p.price)
````

## 15. Encapsulation + Properties

Python often uses a leading underscore such as `_salary` for internal storage and exposes controlled access through a property.

This provides encapsulation without writing Java-style `get_salary()` and `set_salary()` methods for every field.

````python
class Employee:
    def __init__(self, salary):
        self.salary = salary

    @property
    def salary(self):
        return self._salary

    @salary.setter
    def salary(self, value):
        if value < 0:
            raise ValueError("Salary cannot be negative")
        self._salary = value

e = Employee(50000)
e.salary = 55000
print(e.salary)
````

## 16. Factory Methods

A factory method is a method that creates and returns objects. Class methods are often used for this purpose.

````python
class User:
    def __init__(self, username, role):
        self.username = username
        self.role = role

    @classmethod
    def admin(cls, username):
        return cls(username, "admin")

    @classmethod
    def guest(cls, username):
        return cls(username, "guest")

print(User.admin("ajim").role)
print(User.guest("visitor").role)
````

## 17. Practical Example — Data Science Configuration

Class methods can provide predefined configurations while static methods can validate independent values.

````python
class MLConfig:
    default_test_size = 0.2

    def __init__(self, test_size, random_state):
        self.test_size = test_size
        self.random_state = random_state

    @classmethod
    def default(cls):
        return cls(cls.default_test_size, 42)

    @staticmethod
    def valid_test_size(value):
        return 0 < value < 1

config = MLConfig.default()
print(config.test_size, config.random_state)
print(MLConfig.valid_test_size(0.2))
````

## 18. Practical Example — Model Wrapper

A property can expose model status as a computed/read-only value, while a class method can provide a standard constructor.

````python
class Model:
    def __init__(self, name, trained=False):
        self.name = name
        self._trained = trained

    @property
    def is_trained(self):
        return self._trained

    def train(self):
        self._trained = True

    @classmethod
    def pretrained(cls, name):
        return cls(name, trained=True)

model = Model("Logistic Regression")
print(model.is_trained)
model.train()
print(model.is_trained)
print(Model.pretrained("ResNet").is_trained)
````

## 19. Common Mistakes

- Using `@classmethod` when object-specific state is required
- Using `@staticmethod` when class/object state is actually needed
- Forgetting `cls` in a class method
- Writing unnecessary getters/setters instead of Python properties
- Storing directly into the public property inside its own setter, causing recursion
- Using a property for trivial data where direct attribute access is clearer

## 27. Final Mental Model

```text
INSTANCE METHOD
    self → object-specific state

CLASS METHOD
    cls → class-specific state / alternative constructors

STATIC METHOD
    no self / cls → related utility logic

PROPERTY
    attribute syntax → method logic / validation / computed values
```

### Key Rule
Choose the method type based on **what data the behavior needs**.
